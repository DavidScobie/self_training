```python
#####exercise 1
class Car:
    def __init__(self,color,mileage): #method (constructor)
        self.color = color
        self.mileage = mileage

    def __str__(self):  #_str_ is a dunder method
        return f"The {self.color} car has {self.mileage} miles"

car1 = Car(color="blue", mileage=20_000)
car2 = Car(color="red", mileage=30_000)

print(car1)
print(car2)
```

More advanced examples

```python
class Dog:     #this is our dog class
    species = "Canis familiaris"    #class atrribute

    def __init__(self, name, age):
        self.name = name   #instance attribute
        self.age = age   #instance attribute

    def __str__(self):
        return f"{self.name} is {self.age} years old"

    def speak(self, sound): 
        return f"{self.name} says {sound}" #notice that sound isnt an attribute

class Bulldog(Dog): #the Bulldog is the child class of the Dog parent class
    pass 

jack = Bulldog("Jack", 3)

print(jack.name)
print(jack.species) #
print(isinstance(jack, Dog)) #check if jack is from the Dog class
print(jack.speak("woof"))
jack.age = 4 #we can reassign a value to an instance attribute
print(jack)

class JackRussellTerrier(Dog):
    def speak(self, sound="Arf"): #override a method on the parent class
        return f"{self.name} makes the noise: {sound}" 

miles = JackRussellTerrier("Miles", 4)
print(miles.speak())
print(miles.speak("grrr")) #custom objects are 'mutable' by default (we can alter them dynamically)
```

```python
class GoldenRetriever(Dog):
    def speak(self, sound="Bark"):
        return super().speak(sound) #access the parent class from inside a method of the child class

max = GoldenRetriever(name="max",age=5)
print(max.speak())
```

encapsulation example https://www.programiz.com/python-programming/object-oriented-programming


```python
class Computer:

    def __init__(self):
        self.__maxprice = 900

    def sell(self):
        print("Selling Price: {}".format(self.__maxprice))

    def setMaxPrice(self, price):
        self.__maxprice = price

c = Computer()
c.sell()

#we cant change the price this way, because __maxprice is a private variable
c.__maxprice = 1000
c.sell()

# using a setter function, we can change the price
c.setMaxPrice(1000)
c.sell()
```

This is an example from a google search (train, but different scenario)

```python
class Passenger:
    def __init__(self, name, seat_number, destination):
        self.name = name
        self.seat_number = seat_number
        self.destination = destination

    def __str__(self):
        return f"Passenger: {self.name}, Seat: {self.seat_number}, Destination: {self.destination}"


class Train:
    def __init__(self, train_id, destination):
        self.train_id = train_id
        self.destination = destination
        self.passengers = []

    def add_passenger(self, passenger):
        self.passengers.append(passenger)

    def list_passengers(self):
        for passenger in self.passengers:
            print(passenger)

    def __str__(self):
        return f"Train {self.train_id} to {self.destination} with {len(self.passengers)} passengers"


# Example usage
train = Train("T123", "New York")
passenger1 = Passenger("Alice", "A12", "New York")
passenger2 = Passenger("Bob", "B24", "New York")
train.add_passenger(passenger1)
train.add_passenger(passenger2)

print(train)
train.list_passengers()
```

a more refined online example

```python

class Passenger:
    def __init__(self, mass, luggage_mass):
        self.mass = mass
        self.luggage_mass = luggage_mass

    def total_mass(self):
        return self.mass + self.luggage_mass


class Carriage:
    def __init__(self, name, mass):
        self.name = name
        self.mass = mass
        self.passengers = []

    def add_passenger(self, passenger):
        self.passengers.append(passenger)

    def total_mass(self):
        passenger_total = sum(p.total_mass() for p in self.passengers)
        return self.mass + passenger_total


class Train:
    def __init__(self):
        self.carriages = []

    def add_carriage(self, carriage):
        self.carriages.append(carriage)

    def total_mass(self):
        return sum(c.total_mass() for c in self.carriages)

    def carriage_masses(self):
        return {c.name: c.total_mass() for c in self.carriages}
        
# Create passengers
p1 = Passenger(70, 10)
p2 = Passenger(60, 5)
p3 = Passenger(80, 15)

# Create carriages
cA = Carriage("A", 1000)
cB = Carriage("B", 1200)

# Add passengers to carriages
cA.add_passenger(p1)
cA.add_passenger(p2)
cB.add_passenger(p3)

# Create train and add carriages
train = Train()
train.add_carriage(cA)
train.add_carriage(cB)

# Results
print("Total train mass:", train.total_mass())
print("Mass per carriage:", train.carriage_masses())
```
